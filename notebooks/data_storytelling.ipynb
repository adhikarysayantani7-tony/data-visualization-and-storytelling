## 1. Setup & Load Data


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('../data/ecommerce_story_data.csv')
df.head()
```

## 2. Story Part 1: Revenue Tripled From Jan ($42K) to Dec ($128K)
A line chart with shaded area and callout annotation highlights the Q4 holiday surge.


```python
plt.figure(figsize=(9, 4.5))
plt.plot(df['Month'], df['Revenue_USD_K'], marker='o', linewidth=2.8, color='#2b6cb0', label='Monthly Revenue ($K)')
plt.fill_between(df['Month'], df['Revenue_USD_K'], color='#2b6cb0', alpha=0.12)
plt.annotate('Holiday Surge\n($128K in Dec)', xy=('Dec', 128), xytext=('Sep', 115),
             arrowprops=dict(arrowstyle='->', color='#e53e3e', lw=1.8),
             fontsize=10, fontweight='bold', color='#e53e3e')
plt.title('Story Part 1: Revenue Tripled From Jan ($42K) to Dec ($128K)', fontsize=12, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Revenue ($ Thousands)')
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

## 3. Story Part 2: The Mobile Conversion Gap
Grouped bar chart comparing Desktop vs. Mobile conversion rates across all 12 months.


```python
plt.figure(figsize=(9, 4.5))
x = np.arange(len(df['Month']))
w = 0.36
plt.bar(x - w/2, df['Desktop_Conv_Pct'], width=w, color='#1a365d', label='Desktop Conversion (%)')
plt.bar(x + w/2, df['Mobile_Conv_Pct'], width=w, color='#ed8936', label='Mobile Conversion (%)')
plt.xticks(x, df['Month'])
plt.title('Story Part 2: Desktop Converts ~1.5% Higher Than Mobile Every Month', fontsize=12, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Conversion Rate (%)')
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. Story Part 3: Regional Revenue Breakdown
Horizontal bar chart emphasizing the dominant share of North America and Europe.


```python
regions = ['North America', 'Europe', 'Asia-Pacific', 'Latin America']
reg_rev = [365, 240, 185, 78]
plt.figure(figsize=(8.5, 4))
bars = plt.barh(regions[::-1], reg_rev[::-1], color=['#cbd5e0', '#63b3ed', '#3182ce', '#1a365d'])
for b in bars:
    plt.text(b.get_width() + 6, b.get_y() + b.get_height()/2, f"${int(b.get_width())}K", va='center', fontweight='bold')
plt.xlim(0, 420)
plt.title('Story Part 3: North America & Europe Drive 70% of Annual Revenue', fontsize=12, fontweight='bold')
plt.xlabel('Annual Revenue ($ Thousands)')
plt.tight_layout()
plt.show()
```

## 5. Story Part 4: Protecting Profit Margins From Heavy Discounts
Regression scatter plot identifying the 20% discount threshold.


```python
discounts = np.array([0, 5, 10, 12, 15, 18, 20, 25, 30, 35, 40])
margins = np.array([34, 33, 31, 29, 27, 24, 22, 17, 12, 7, 3])
plt.figure(figsize=(8.5, 4))
sns.regplot(x=discounts, y=margins, color='#e53e3e', scatter_kws={'s': 70})
plt.axvline(20, color='#2f855a', linestyle='--', label='Optimal Discount Cap (20%)')
plt.title('Story Part 4: Discounts Above 20% Erode Profit Margins Below 22%', fontsize=12, fontweight='bold')
plt.xlabel('Discount Offered (%)')
plt.ylabel('Net Profit Margin (%)')
plt.legend()
plt.tight_layout()
plt.show()
```
